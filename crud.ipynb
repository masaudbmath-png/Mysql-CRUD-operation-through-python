```python
import mysql.connector
```

```python
try:
    conn = mysql.connector.connect(
        host='127.0.0.1',
        user='root',  # your database user name
        password='abc@123', # your database password
        database='airportdb' # the database your working on
    )

    mycursor = conn.cursor()   # cursor method allow us to execute sql quries
    print('Connection established successfully')

except:
    print('Connection error')
```

```python
mycursor.execute("CREATE DATABASE airportdb")  # this will create a new database named aiportdb on server
conn.commit()
print('database created succesfully')
print('*'*35)
```

```python
# creating table

mycursor.execute('use airportdb')
mycursor.execute("""
    create table airport (airport_id integer primary key,
    code varchar(10) not null,
    city varchar(50) not null,
    name varchar(100) not null
    )
    """)
conn.commit()
print("Airport table created successfully")
print('*'*35)
```

```python
#inserting data into table
mycursor.execute("""
    INSERT INTO airport (airport_id, code, city, name)
    VALUES
        (1, 'DEL', 'isb', 'Islamabad'),
        (2, 'BKK', 'pesh', 'Peshawar')
""")

conn.commit()
```

```python
mycursor.execute("SELECT * FROM airport")

data = mycursor.fetchall()

for row in data:
    print(row[3])
```

```python
mycursor.execute('delete from airport where airport_id=1')
conn.commit()
print('*'*35)
print("After deletion")
print('*'*35)
mycursor.execute("SELECT * FROM airport")

data = mycursor.fetchall()

for row in data:
    print(row[3])
```
